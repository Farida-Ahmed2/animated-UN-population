```python
import pandas as pd

df = pd.read_csv('C:/Users/user/Documents/Time_series_in python/data/un-country-data.csv')

df.head(5)
```

```python
countries = df[df['ISO3_code'].notnull()]
#countries.columns

columns = ['ISO3_code','Location','Time','TPopulation1Jan','TPopulation1July']
clean_df = countries[columns]
clean_df['TPopulation1Jan'] = clean_df['TPopulation1Jan']/1000
clean_df['TPopulation1July'] = clean_df['TPopulation1July']/1000

clean_df.head()
clean_df.to_csv('C:/Users/user/Documents/Time_series_in python/data/cleaned_country_data.csv', index=False)
```
